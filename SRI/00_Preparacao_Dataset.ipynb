```
# Carrega o dataset de tweets em português do Hugging Face
from datasets import load_dataset

dataset = load_dataset("miguelribeirokk/crime_tweets_in_portuguese")
```

```
# Converte o dataset em um DataFrame (tabela) e mostra as primeiras linhas
import pandas as pd

df = pd.DataFrame(dataset['train'])
df.head()
```

```
# Seleciona a coluna de texto (tweets), renomeia para "documento",
df_documentos = df[['text']].copy()

df_documentos.columns = ['documento']
df_documentos.insert(0, 'id', range(1, len(df_documentos) + 1))

df_documentos = df_documentos.head(50)

df_documentos.head()
```
